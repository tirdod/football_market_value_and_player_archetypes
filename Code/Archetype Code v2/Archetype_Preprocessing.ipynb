```python
import pandas as pd
import numpy as np

def generate_per90_and_total(df):
    """
    Generates per90 and total datasets from input DataFrame using 'minutes_played',
    skipping columns that already contain '/90' in their name.

    Parameters:
        df (pd.DataFrame): The input DataFrame, must include 'minutes_played' column.

    Returns:
        tuple: (per90_df, total_df)
    """
    # Columns to exclude from per90 generation
    exclude_keywords = ['/sh', '/sot', '%', '/90']
    exclude_columns = [col for col in df.columns if any(keyword in col.lower() for keyword in exclude_keywords)]

    # Non-divisible base columns (categorical or already used)
    non_divisible = ['season', 'league', 'team', 'player', 'nation', 'average_shot_distance_meters', 'minutes_played','90s','shots_on_target_percentage','goals_per_shot','goals_per_shot_on_target']

    # Columns eligible for per90 calculation
    columns_to_convert = [col for col in df.columns if col not in (non_divisible + exclude_columns)]

    # Ensure no divide-by-zero
    df = df.copy()
    df['minutes_played'] = df['minutes_played'].replace(0, np.nan)

    # Create per90 DataFrame
    per90_df = df.copy()
    for col in columns_to_convert:
        per90_df[col] = per90_df[col] / (per90_df['minutes_played'] / 90)

    # Rename converted columns
    rename_map = {col: f"{col}_per90" for col in columns_to_convert}
    per90_df.rename(columns=rename_map, inplace=True)

    # Round and drop 'mp' from per90 frame
    per90_df[list(rename_map.values())] = per90_df[list(rename_map.values())].round(2)
    per90_df.drop(columns='minutes_played', inplace=True)

    # Total dataset: drop existing '/90' columns to keep only raw values
    total_df = df.drop(columns=[col for col in df.columns if '90' in col.lower()], errors='ignore')

    return per90_df, total_df
```

```python
def generate_per90_and_total_passing(df):
    """
    Generates per90 and total datasets from passing DataFrame using 'mp' (minutes played),
    skipping columns that are percentages or already per-90.

    Parameters:
        df (pd.DataFrame): Input passing DataFrame with 'mp' column.

    Returns:
        tuple: (per90_df, total_df)
    """
    # Keywords that indicate we should exclude the column from per90 conversion
    exclude_keywords = ['%', '/90']  # already rate-based or percentages
    exclude_columns = [col for col in df.columns if any(keyword in col.lower() for keyword in exclude_keywords)]

    # Non-divisible base columns (categorical or already used)
    non_divisible = ['season', 'league', 'team', 'player', 'minutes_played', '90s']

    # Columns to convert to per90
    columns_to_convert = [col for col in df.columns if col not in (non_divisible + exclude_columns)]

    # Make a copy and prevent divide-by-zero
    df = df.copy()
    df['minutes_played'] = df['minutes_played'].replace(0, np.nan)

    # Create per90 DataFrame
    per90_df = df.copy()
    for col in columns_to_convert:
        per90_df[col] = per90_df[col] / (per90_df['minutes_played'] / 90)

    # Rename columns to indicate per90
    rename_map = {col: f"{col}_per90" for col in columns_to_convert}
    per90_df.rename(columns=rename_map, inplace=True)

    # Round per90 values
    per90_df[list(rename_map.values())] = per90_df[list(rename_map.values())].round(2)

    # Drop raw minute column from per90 frame
    per90_df.drop(columns='minutes_played', inplace=True)

    # Total DataFrame: keep raw columns and drop those with '/90' if any
    total_df = df.drop(columns=[col for col in df.columns if '90' in col.lower()], errors='ignore')

    return per90_df, total_df
```

```python
def generate_per90_and_total_goal_and_shot_creation(df):
    """
    Generates per90 and total datasets for goal and shot creation stats using 'mp'.

    Parameters:
        df (pd.DataFrame): DataFrame including 'mp' and raw event counts.

    Returns:
        tuple: (per90_df, total_df)
    """
    # Keywords to exclude from per90 transformation
    exclude_keywords = ['sca90', 'gca90']  # already per90 columns
    exclude_columns = [col for col in df.columns if any(keyword in col.lower() for keyword in exclude_keywords)]

    # Columns that should not be divided
    non_divisible = ['season', 'league', 'team', 'player', 'minutes_played', '90s']

    # Columns eligible for per90 conversion
    columns_to_convert = [col for col in df.columns if col not in (non_divisible + exclude_columns)]

    # Avoid divide-by-zero
    df = df.copy()
    df['minutes_played'] = df['minutes_played'].replace(0, np.nan)

    # Create per90 DataFrame
    per90_df = df.copy()
    for col in columns_to_convert:
        per90_df[col] = per90_df[col] / (per90_df['minutes_played'] / 90)

    # Rename newly converted columns
    rename_map = {col: f"{col}_per90" for col in columns_to_convert}
    per90_df.rename(columns=rename_map, inplace=True)

    # Round and drop 'mp' from per90 frame
    per90_df[list(rename_map.values())] = per90_df[list(rename_map.values())].round(2)
    per90_df.drop(columns='minutes_played', inplace=True)

    # Total frame drops per90 columns
    total_df = df.drop(columns=[col for col in df.columns if '90' in col.lower()], errors='ignore')

    return per90_df, total_df
```

```python
def generate_per90_and_total_defensive_actions(df):
    """
    Generates per90 and total datasets for defensive actions data using 'minutes_played',
    excluding percentage and metadata columns.

    Parameters:
        df (pd.DataFrame): Defensive actions DataFrame with 'minutes_played' and '90s'.

    Returns:
        tuple: (per90_df, total_df)
    """
    df = df.copy()

    # Avoid divide-by-zero issues
    df['minutes_played'] = df['minutes_played'].replace(0, np.nan)

    # Define exclude_cols by name
    exclude_by_name = ['season', 'league', 'team', 'player', 'minutes_played', '90s']

    # Exclude columns that contain '%' or 'percentage'
    exclude_by_keyword = [col for col in df.columns if '%' in col.lower() or 'percentage' in col.lower()]

    # Final set of columns to exclude
    exclude_cols = set(exclude_by_name + exclude_by_keyword)

    # Columns to convert = everything else
    columns_to_convert = [col for col in df.columns if col not in exclude_cols]

    # Create per90 DataFrame
    per90_df = df.copy()
    for col in columns_to_convert:
        per90_df[col] = (per90_df[col] / (per90_df['minutes_played'] / 90)).fillna(0)

    # Rename converted columns
    per90_df.rename(columns={col: f"{col}_per90" for col in columns_to_convert}, inplace=True)

    # Drop raw minutes and round
    per90_df.drop(columns='minutes_played', inplace=True)
    per90_df = per90_df.round(2)

    # Total dataset (raw values)
    total_df = df.drop(columns=['90s'], errors='ignore')

    return per90_df, total_df
```

```python
def generate_per90_and_total_possession(df):
    """
    Generates per90 and total datasets from possession DataFrame using 'minutes_played',
    skipping percentage columns and already per90 columns.

    Parameters:
        df (pd.DataFrame): The possession DataFrame (must include 'minutes_played').

    Returns:
        tuple: (per90_df, total_df)
    """
    df = df.copy()

    # Drop columns that are already in per90 format
    per_90_existing_columns = [col for col in df.columns if '/90' in col.lower()]
    df.drop(columns=per_90_existing_columns, inplace=True, errors='ignore')

    # Replace 0 with NaN to avoid divide-by-zero errors
    df['minutes_played'] = df['minutes_played'].replace(0, np.nan)

    # Exclude columns with 'percentage' and metadata
    exclude_columns = [col for col in df.columns if 'percentage' in col.lower()]
    non_divisible_columns = ['season', 'league', 'team', 'player', 'minutes_played'] + exclude_columns

    # Identify columns to convert
    columns_to_divide = [col for col in df.columns if col not in non_divisible_columns]

    # Create per90 DataFrame
    per90_df = df.copy()
    for col in columns_to_divide:
        per90_df[col] = (per90_df[col] / (per90_df['minutes_played'] / 90)).fillna(0)

    # Rename per90 columns
    rename_columns = {col: f"{col}_per90" for col in columns_to_divide}
    per90_df.rename(columns=rename_columns, inplace=True)

    # Round values
    per90_df[list(rename_columns.values())] = per90_df[list(rename_columns.values())].round(2)

    # Drop raw 'minutes_played' from per90
    per90_df.drop(columns={'minutes_played', '90s_per90'}, inplace=True)

    # Total df is just original minus per90 columns
    total_df = df.drop(columns=['minutes_played','90s'], errors='ignore')

    return per90_df, total_df
```
